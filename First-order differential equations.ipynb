## 1. Introduction

**A differential equation is an equation that contains a derivative of an initially unknown function $f(x)$. It is solved by finding a function $f(x)$ that satisfies the equation.**

A simple example is below:

$\dfrac{df}{dx} = -Af(x)$, where $A$ > 0 is a constant

**It is first-order because it only contains the first derivative**: $\dfrac{df}{dx}$ as opposed to a second-order which would resemble something like this $\dfrac{d^2f}{dx^2}$.

By inspection alone this simple first-order differential equation can be solved. The right side is the result of a function $f(x)$ that has been differentiated so we wanna reverse back and get that function. The left hand side is a rate of change (or gradient) that is proportional to the right hand side. A potential candidate for finding the unknown function is the exponential function $e^x$. 

In fact when a rate of decrease is proportional to the magnitude of that quantity that is decreasing that is easily representted by a decay factor: $e^{-ax}$.

Intuition shows that the required function for the differential equation above is simply: $f(x) = e^{-Ax}$. Differentiating $f(x)$ is:  $\dfrac{df}{dx} = -Af(x)$

## 1.1 A simple example of contructing a differential equation

A scientist wants to construct a model of light penetrating into the ocean. She thinks to herself that the deeper one goes into the depths of the ocean the less intense the light intensity becomes. She wants to find a function $I(x)$ to model this light intensity with regards to ocean depth. 

She reasons that going down in depth by a small change in $x$ (distance) with $x + δx$ the light intensity ($I$) should reduce by a small amount $I + δI$. She expects that the light absorbed by a thin layer is expected to be proportional to the thickness of the layer so that: $δI = −AI δx$ with $A$ being the constant of proportionality.
